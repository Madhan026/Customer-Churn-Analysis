```python
import pandas as pd
import pyodbc

server = r'localhost\SQLEXPRESS'
database = 'ChurnDB'

connection_string = (
    'DRIVER={ODBC Driver 17 for SQL Server};'
    f'SERVER={server};'
    f'DATABASE={database};'
    'Trusted_Connection=yes;'
    'TrustServerCertificate=yes;'
)

conn = pyodbc.connect(connection_string)

query = "SELECT * FROM dbo.vw_ChurnData"

df = pd.read_sql(query, conn)

conn.close()

print("Data loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
df_clean = df.drop(columns=['CustomerID']).copy()
```

```python
df_clean.shape
```

```python
yes_no_columns = [
    'Partner',
    'Dependents',
    'PhoneService',
    'PaperlessBilling',
    'Churn'
]

for col in yes_no_columns:
    df_clean[col] = df_clean[col].map({'Yes': 1, 'No': 0})
```

```python
df_clean[yes_no_columns].head()
```

```python
categorical_columns = [
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaymentMethod'
]

df_clean = pd.get_dummies(
    df_clean,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)
```

```python
print(df_clean.head())
```

```python
print(df_clean.shape)
df_clean.info()
```

```python
print(df_clean.isnull().sum())
```

```python
print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
output_file = 'cleaned_churn_data.csv'

df_clean.to_csv(output_file, index=False)

print(f"Cleaned dataset saved successfully as {output_file}")
```

```python
df_clean['TotalServicesUsed'] = (
    df_clean['PhoneService']
    + df_clean['MultipleLines_Yes']
    + (1 - df_clean['InternetService_No'])
    + df_clean['OnlineSecurity_Yes']
    + df_clean['OnlineBackup_Yes']
    + df_clean['DeviceProtection_Yes']
    + df_clean['TechSupport_Yes']
    + df_clean['StreamingTV_Yes']
    + df_clean['StreamingMovies_Yes']
)
```

```python
df_clean[['TotalServicesUsed', 'Churn']].head(10)
```

```python
import numpy as np

df_clean['AvgMonthlySpend'] = np.where(
    df_clean['tenure'] > 0,
    df_clean['TotalCharges'] / df_clean['tenure'],
    df_clean['MonthlyCharges']
)
```

```python
df_clean[['tenure', 'TotalCharges', 'MonthlyCharges', 'AvgMonthlySpend']].head(10)
```

```python
correlation = df_clean[
    ['TotalServicesUsed', 'AvgMonthlySpend', 'Churn']
].corr()

print(correlation)
```

```python
print("TotalServicesUsed vs Churn:",
      df_clean['TotalServicesUsed'].corr(df_clean['Churn']))

print("AvgMonthlySpend vs Churn:",
      df_clean['AvgMonthlySpend'].corr(df_clean['Churn']))
```
